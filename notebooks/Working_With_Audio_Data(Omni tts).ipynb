```python
import torch, soundfile as sf, os
from omnivoice import OmniVoice
from IPython.display import Audio

model = OmniVoice.from_pretrained("k2-fsa/OmniVoice", device_map="cuda:0", dtype=torch.float16,)


```

```python
def generate_bangla_audio(text, filename, instruct="female", out_dir="/content/outputs"):
    os.makedirs(out_dir, exist_ok=True)
    audio = model.generate(text=text, instruct=instruct)
    path = f"{out_dir}/{filename}.wav"
    sf.write(path, audio[0], 24000)
    return path

```

Generate a first voice (to be cloned)

```python
p = generate_bangla_audio("আমার সোনার বাংলা, আমি তোমায় ভালোবাসি।", "reference1")
Audio(p)
```

```python
def generate_bangla_audio_cloned(text, filename, ref_audio="/content/outputs/male voice.wav", ref_text="একাধিক শব্দ পাশাপাশি বসে যখন সম্পূর্ণ মনের ভাব প্রকাশ করে, তখন তাকে সেন্টেন্স বা বাক্য বলে।", out_dir="/content/outputs"):
    os.makedirs(out_dir, exist_ok=True)
    # If ref_audio exists, we clone its speaker style, otherwise we fallback to generic voice design
    if os.path.exists(ref_audio):
        audio = model.generate(text=text, ref_text=ref_text, ref_audio=ref_audio,instruct="female")
    else:
        # Fallback to create the first reference audio if it doesn't exist yet
        audio = model.generate(text=text, instruct="female")
    path = f"{out_dir}/{filename}.wav"
    sf.write(path, audio[0], 24000)
    return path
```

```python
p3 = generate_bangla_audio_cloned("আমরা শিগ্রই আপনার সাথে যোগাযোগ করবো ", "test3")
Audio(p3)
```
